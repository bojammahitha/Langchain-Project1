```python
# ============================================================
# E-COMMERCE AI DATA ANALYST
# Single-cell Google Colab project
#
# Features:
#   - Creates realistic e-commerce data
#   - SQLite database
#   - LangChain SQL Agent
#   - Natural-language questions
#   - SQL generation + execution
#   - Business analysis
#   - Safety checks for SQL
#   - Interactive chat
# ============================================================

# =========================
# 1. INSTALL DEPENDENCIES
# =========================

!pip -q install -U langchain langchain-openai langchain-community pandas matplotlib faker


# =========================
# 2. IMPORTS
# =========================

import os
import re
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt

from faker import Faker
from getpass import getpass

from langchain_openai import ChatOpenAI
from langchain_community.utilities import SQLDatabase
from langchain_community.agent_toolkits import create_sql_agent


# =========================
# 3. OPENAI API KEY
# =========================

if "OPENAI_API_KEY" not in os.environ:
    os.environ["OPENAI_API_KEY"] = getpass(
        "Enter your OpenAI API key: "
    )


# =========================
# 4. CREATE DATABASE
# =========================

DB_PATH = "/content/ecommerce.db"

# Remove old database if it exists
if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()

# Customers
cursor.execute("""
CREATE TABLE customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    email TEXT,
    city TEXT,
    state TEXT,
    created_at TEXT
)
""")

# Products
cursor.execute("""
CREATE TABLE products (
    id INTEGER PRIMARY KEY,
    name TEXT,
    category TEXT,
    price REAL,
    cost REAL,
    stock INTEGER
)
""")

# Orders
cursor.execute("""
CREATE TABLE orders (
    id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    order_date TEXT,
    status TEXT,
    total_amount REAL,
    FOREIGN KEY(customer_id) REFERENCES customers(id)
)
""")

# Order items
cursor.execute("""
CREATE TABLE order_items (
    id INTEGER PRIMARY KEY,
    order_id INTEGER,
    product_id INTEGER,
    quantity INTEGER,
    unit_price REAL,
    FOREIGN KEY(order_id) REFERENCES orders(id),
    FOREIGN KEY(product_id) REFERENCES products(id)
)
""")

# Reviews
cursor.execute("""
CREATE TABLE reviews (
    id INTEGER PRIMARY KEY,
    product_id INTEGER,
    customer_id INTEGER,
    rating INTEGER,
    review_text TEXT,
    review_date TEXT,
    FOREIGN KEY(product_id) REFERENCES products(id),
    FOREIGN KEY(customer_id) REFERENCES customers(id)
)
""")


# =========================
# 5. GENERATE SAMPLE DATA
# =========================

fake = Faker("en_IN")

import random
from datetime import datetime, timedelta

random.seed(42)
Faker.seed(42)

# ---------- Customers ----------

indian_cities = [
    ("Bengaluru", "Karnataka"),
    ("Mysuru", "Karnataka"),
    ("Mumbai", "Maharashtra"),
    ("Pune", "Maharashtra"),
    ("Delhi", "Delhi"),
    ("Hyderabad", "Telangana"),
    ("Chennai", "Tamil Nadu"),
    ("Kochi", "Kerala"),
    ("Ahmedabad", "Gujarat"),
    ("Jaipur", "Rajasthan"),
]

customers = []

for i in range(1, 501):

    city, state = random.choice(indian_cities)

    customers.append((
        i,
        fake.name(),
        fake.email(),
        city,
        state,
        (
            datetime.now() -
            timedelta(days=random.randint(30, 1000))
        ).strftime("%Y-%m-%d")
    ))

cursor.executemany("""
INSERT INTO customers
(id, name, email, city, state, created_at)
VALUES (?, ?, ?, ?, ?, ?)
""", customers)


# ---------- Products ----------

product_catalog = [
    ("Laptop Pro 14", "Electronics", 74999, 60000),
    ("Laptop Air 13", "Electronics", 64999, 51000),
    ("Gaming Laptop X", "Electronics", 89999, 72000),
    ("Wireless Mouse", "Electronics", 1499, 800),
    ("Mechanical Keyboard", "Electronics", 3499, 2100),
    ("Noise Cancelling Headphones", "Electronics", 7999, 5200),
    ("Bluetooth Speaker", "Electronics", 3999, 2500),
    ("Smart Watch", "Electronics", 5999, 3800),
    ("USB-C Hub", "Electronics", 2499, 1400),
    ("Power Bank", "Electronics", 1999, 1100),

    ("Running Shoes", "Fashion", 4999, 2800),
    ("Casual Sneakers", "Fashion", 3999, 2300),
    ("Formal Shoes", "Fashion", 5499, 3200),
    ("Denim Jacket", "Fashion", 3499, 2000),
    ("Cotton T-Shirt", "Fashion", 999, 450),
    ("Hoodie", "Fashion", 1999, 1000),

    ("Coffee Maker", "Home", 6999, 4300),
    ("Air Fryer", "Home", 8999, 5800),
    ("Mixer Grinder", "Home", 5499, 3400),
    ("Electric Kettle", "Home", 1999, 1100),
    ("Vacuum Cleaner", "Home", 11999, 7800),
    ("Desk Lamp", "Home", 1499, 700),

    ("Programming in Python", "Books", 799, 400),
    ("Data Science Handbook", "Books", 1299, 700),
    ("AI Engineering", "Books", 999, 550),
    ("Clean Code", "Books", 699, 350),
]

products = []

for i, (name, category, price, cost) in enumerate(product_catalog, 1):

    stock = random.randint(5, 250)

    products.append((
        i,
        name,
        category,
        price,
        cost,
        stock
    ))

cursor.executemany("""
INSERT INTO products
(id, name, category, price, cost, stock)
VALUES (?, ?, ?, ?, ?, ?)
""", products)


# ---------- Orders + Order Items ----------

order_records = []
order_item_records = []

order_id = 1
order_item_id = 1

product_prices = {
    p[0]: p[3] for p in products
}

for _ in range(2500):

    customer_id = random.randint(1, 500)

    order_date = (
        datetime.now() -
        timedelta(days=random.randint(0, 365))
    ).strftime("%Y-%m-%d")

    status = random.choices(
        ["completed", "cancelled", "returned", "pending"],
        weights=[80, 5, 10, 5],
        k=1
    )[0]

    number_of_products = random.randint(1, 4)

    selected_products = random.sample(
        list(product_prices.keys()),
        number_of_products
    )

    total_amount = 0

    temporary_items = []

    for product_id in selected_products:

        quantity = random.randint(1, 3)

        unit_price = product_prices[product_id]

        # Random discount
        discount = random.choice([
            0,
            0,
            0,
            0.05,
            0.10,
            0.15
        ])

        final_price = round(
            unit_price * (1 - discount),
            2
        )

        total_amount += quantity * final_price

        temporary_items.append((
            order_item_id,
            order_id,
            product_id,
            quantity,
            final_price
        ))

        order_item_id += 1

    order_records.append((
        order_id,
        customer_id,
        order_date,
        status,
        round(total_amount, 2)
    ))

    order_item_records.extend(temporary_items)

    order_id += 1


cursor.executemany("""
INSERT INTO orders
(id, customer_id, order_date, status, total_amount)
VALUES (?, ?, ?, ?, ?)
""", order_records)

cursor.executemany("""
INSERT INTO order_items
(id, order_id, product_id, quantity, unit_price)
VALUES (?, ?, ?, ?, ?)
""", order_item_records)


# ---------- Reviews ----------

review_texts = {
    5: [
        "Excellent product. Highly recommended.",
        "Amazing quality and great value.",
        "Very happy with the purchase.",
        "Works perfectly.",
    ],
    4: [
        "Good product and worth the money.",
        "Very good overall.",
        "Quality is good.",
        "Happy with the purchase.",
    ],
    3: [
        "Average product.",
        "It is okay for the price.",
        "Could be improved.",
        "Decent but not exceptional.",
    ],
    2: [
        "Not very satisfied.",
        "Quality could be better.",
        "Expected more from this product.",
    ],
    1: [
        "Very disappointing.",
        "Product stopped working quickly.",
        "Not worth the price.",
    ],
}

review_records = []

for review_id in range(1, 2501):

    product_id = random.randint(1, len(products))
    customer_id = random.randint(1, 500)

    rating = random.choices(
        [1, 2, 3, 4, 5],
        weights=[5, 5, 15, 30, 45],
        k=1
    )[0]

    review = random.choice(
        review_texts[rating]
    )

    review_date = (
        datetime.now() -
        timedelta(days=random.randint(0, 365))
    ).strftime("%Y-%m-%d")

    review_records.append((
        review_id,
        product_id,
        customer_id,
        rating,
        review,
        review_date
    ))

cursor.executemany("""
INSERT INTO reviews
(id, product_id, customer_id, rating, review_text, review_date)
VALUES (?, ?, ?, ?, ?, ?)
""", review_records)


conn.commit()


# =========================
# 6. DATABASE SUMMARY
# =========================

print("==========================================")
print(" E-COMMERCE DATABASE CREATED")
print("==========================================")

tables = [
    "customers",
    "products",
    "orders",
    "order_items",
    "reviews"
]

for table in tables:

    count = pd.read_sql(
        f"SELECT COUNT(*) AS count FROM {table}",
        conn
    ).iloc[0]["count"]

    print(f"{table:15} : {count:,} records")


# =========================
# 7. CONNECT LANGCHAIN
# =========================

db = SQLDatabase.from_uri(
    f"sqlite:///{DB_PATH}"
)

llm = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0
)


# =========================
# 8. CREATE SQL AGENT
# =========================

agent = create_sql_agent(
    llm=llm,
    db=db,
    agent_type="tool-calling",
    verbose=False,
    prefix="""
You are an expert e-commerce data analyst.

Your job is to answer business questions using the SQLite
e-commerce database.

Important rules:

1. Inspect the database schema before answering.
2. Generate SQL queries when database information is required.
3. Only perform READ operations.
4. Never INSERT, UPDATE, DELETE, DROP, ALTER, or modify data.
5. Do not invent numbers.
6. Base numerical claims on query results.
7. Explain results in simple business language.
8. When useful, provide percentages and comparisons.
9. If the question cannot be answered from the database,
   clearly say so.
10. For revenue calculations, use:
       quantity * unit_price
    from order_items.
11. For profit calculations, use:
       (unit_price - product.cost) * quantity
12. Treat "completed" orders as successful sales unless the
    question specifies otherwise.
"""
)


# =========================
# 9. SAFE QUESTION FUNCTION
# =========================

def ask_analyst(question):

    dangerous_keywords = [
        "DROP ",
        "DELETE ",
        "UPDATE ",
        "INSERT ",
        "ALTER ",
        "TRUNCATE ",
        "CREATE "
    ]

    upper_question = question.upper()

    for keyword in dangerous_keywords:

        if keyword in upper_question:

            return (
                "For safety, this analyst only supports "
                "read-only business analysis."
            )

    try:

        result = agent.invoke({
            "input": question
        })

        return result["output"]

    except Exception as e:

        return f"Error: {str(e)}"


# =========================
# 10. EXAMPLE QUESTIONS
# =========================

example_questions = [
    "What are the top 5 products by revenue?",
    "Which category generates the most revenue?",
    "What is our total revenue?",
    "Which products have low inventory?",
    "What are the average ratings for each product?",
    "Which city has the most customers?",
    "What are the top 5 products by profit?",
    "How many orders were cancelled?",
]

print("\n==========================================")
print(" EXAMPLE QUESTIONS")
print("==========================================")

for i, q in enumerate(example_questions, 1):

    print(f"{i}. {q}")


# =========================
# 11. INTERACTIVE ANALYST
# =========================

print("\n==========================================")
print(" E-COMMERCE AI ANALYST READY")
print("==========================================")

print("""
Ask questions about your e-commerce data.

Examples:

- What are our top 10 products by revenue?
- Which category has the highest profit?
- What is our cancellation rate?
- Which products have less than 20 units in stock?
- Which city generates the most revenue?
- What are the worst-rated products?
- How much revenue did Electronics generate?
- Which products have high sales but low inventory?

Type 'exit' to stop.
""")

while True:

    question = input("\nYou: ")

    if question.lower().strip() in [
        "exit",
        "quit",
        "q"
    ]:
        print("Goodbye!")
        break

    print("\nAI Analyst:")
    print(ask_analyst(question))
```
