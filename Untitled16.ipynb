````python
# ============================================================
# 🛍️ E-COMMERCE AI DATA ANALYST
# GOOGLE COLAB - SINGLE CELL
# Gemini + LangChain + SQLite
# ============================================================

# ============================================================
# 1. INSTALL
# ============================================================

!pip -q install -U langchain langchain-google-genai faker


# ============================================================
# 2. IMPORTS
# ============================================================

import os
import re
import sqlite3
import random
import pandas as pd

from faker import Faker
from getpass import getpass
from datetime import datetime, timedelta

from langchain_google_genai import ChatGoogleGenerativeAI


# ============================================================
# 3. GEMINI API KEY
# ============================================================

print("=" * 70)
print("🔑 GEMINI API KEY")
print("=" * 70)

if "GEMINI_API_KEY" not in os.environ:
    os.environ["GEMINI_API_KEY"] = getpass(
        "Enter your Gemini API key: "
    )

print("✅ Gemini API key loaded")


# ============================================================
# 4. CREATE SQLITE DATABASE
# ============================================================

DB_PATH = "/content/ecommerce.db"

if os.path.exists(DB_PATH):
    os.remove(DB_PATH)

conn = sqlite3.connect(DB_PATH)
cursor = conn.cursor()


# ============================================================
# 5. CREATE TABLES
# ============================================================

cursor.execute("""
CREATE TABLE customers (
    id INTEGER PRIMARY KEY,
    name TEXT,
    email TEXT,
    city TEXT,
    state TEXT,
    created_at TEXT
)
""")

cursor.execute("""
CREATE TABLE products (
    id INTEGER PRIMARY KEY,
    name TEXT,
    category TEXT,
    price REAL,
    cost REAL,
    stock INTEGER
)
""")

cursor.execute("""
CREATE TABLE orders (
    id INTEGER PRIMARY KEY,
    customer_id INTEGER,
    order_date TEXT,
    status TEXT,
    total_amount REAL,
    FOREIGN KEY(customer_id)
        REFERENCES customers(id)
)
""")

cursor.execute("""
CREATE TABLE order_items (
    id INTEGER PRIMARY KEY,
    order_id INTEGER,
    product_id INTEGER,
    quantity INTEGER,
    unit_price REAL,
    FOREIGN KEY(order_id)
        REFERENCES orders(id),
    FOREIGN KEY(product_id)
        REFERENCES products(id)
)
""")

cursor.execute("""
CREATE TABLE reviews (
    id INTEGER PRIMARY KEY,
    product_id INTEGER,
    customer_id INTEGER,
    rating INTEGER,
    review_text TEXT,
    review_date TEXT,
    FOREIGN KEY(product_id)
        REFERENCES products(id),
    FOREIGN KEY(customer_id)
        REFERENCES customers(id)
)
""")

print("✅ Database tables created")


# ============================================================
# 6. GENERATE DATA
# ============================================================

fake = Faker("en_IN")

random.seed(42)
Faker.seed(42)


# ============================================================
# 7. CUSTOMERS
# ============================================================

indian_cities = [
    ("Bengaluru", "Karnataka"),
    ("Mysuru", "Karnataka"),
    ("Mumbai", "Maharashtra"),
    ("Pune", "Maharashtra"),
    ("Delhi", "Delhi"),
    ("Hyderabad", "Telangana"),
    ("Chennai", "Tamil Nadu"),
    ("Kochi", "Kerala"),
    ("Ahmedabad", "Gujarat"),
    ("Jaipur", "Rajasthan")
]

customers = []

for i in range(1, 501):

    city, state = random.choice(indian_cities)

    created_at = (
        datetime.now()
        - timedelta(days=random.randint(30, 1000))
    ).strftime("%Y-%m-%d")

    customers.append((
        i,
        fake.name(),
        fake.email(),
        city,
        state,
        created_at
    ))

cursor.executemany("""
INSERT INTO customers
(id, name, email, city, state, created_at)
VALUES (?, ?, ?, ?, ?, ?)
""", customers)


# ============================================================
# 8. PRODUCTS
# ============================================================

product_catalog = [

    ("Laptop Pro 14", "Electronics", 74999, 60000),
    ("Laptop Air 13", "Electronics", 64999, 51000),
    ("Gaming Laptop X", "Electronics", 89999, 72000),
    ("Wireless Mouse", "Electronics", 1499, 800),
    ("Mechanical Keyboard", "Electronics", 3499, 2100),
    ("Noise Cancelling Headphones", "Electronics", 7999, 5200),
    ("Bluetooth Speaker", "Electronics", 3999, 2500),
    ("Smart Watch", "Electronics", 5999, 3800),
    ("USB-C Hub", "Electronics", 2499, 1400),
    ("Power Bank", "Electronics", 1999, 1100),

    ("Running Shoes", "Fashion", 4999, 2800),
    ("Casual Sneakers", "Fashion", 3999, 2300),
    ("Formal Shoes", "Fashion", 5499, 3200),
    ("Denim Jacket", "Fashion", 3499, 2000),
    ("Cotton T-Shirt", "Fashion", 999, 450),
    ("Hoodie", "Fashion", 1999, 1000),

    ("Coffee Maker", "Home", 6999, 4300),
    ("Air Fryer", "Home", 8999, 5800),
    ("Mixer Grinder", "Home", 5499, 3400),
    ("Electric Kettle", "Home", 1999, 1100),
    ("Vacuum Cleaner", "Home", 11999, 7800),
    ("Desk Lamp", "Home", 1499, 700),

    ("Programming in Python", "Books", 799, 400),
    ("Data Science Handbook", "Books", 1299, 700),
    ("AI Engineering", "Books", 999, 550),
    ("Clean Code", "Books", 699, 350)
]

products = []

for i, (name, category, price, cost) in enumerate(
    product_catalog,
    start=1
):

    products.append((
        i,
        name,
        category,
        price,
        cost,
        random.randint(5, 250)
    ))

cursor.executemany("""
INSERT INTO products
(id, name, category, price, cost, stock)
VALUES (?, ?, ?, ?, ?, ?)
""", products)


# ============================================================
# 9. ORDERS + ORDER ITEMS
# ============================================================

orders = []
order_items = []

order_id = 1
order_item_id = 1

product_prices = {
    product[0]: product[3]
    for product in products
}

for _ in range(2500):

    customer_id = random.randint(1, 500)

    order_date = (
        datetime.now()
        - timedelta(days=random.randint(0, 365))
    ).strftime("%Y-%m-%d")

    status = random.choices(
        [
            "completed",
            "cancelled",
            "returned",
            "pending"
        ],
        weights=[80, 5, 10, 5],
        k=1
    )[0]

    selected_products = random.sample(
        list(product_prices.keys()),
        random.randint(1, 4)
    )

    total_amount = 0

    for product_id in selected_products:

        quantity = random.randint(1, 3)

        original_price = product_prices[product_id]

        discount = random.choice([
            0,
            0,
            0,
            0.05,
            0.10,
            0.15
        ])

        final_price = round(
            original_price * (1 - discount),
            2
        )

        total_amount += quantity * final_price

        order_items.append((
            order_item_id,
            order_id,
            product_id,
            quantity,
            final_price
        ))

        order_item_id += 1

    orders.append((
        order_id,
        customer_id,
        order_date,
        status,
        round(total_amount, 2)
    ))

    order_id += 1


cursor.executemany("""
INSERT INTO orders
(id, customer_id, order_date, status, total_amount)
VALUES (?, ?, ?, ?, ?)
""", orders)

cursor.executemany("""
INSERT INTO order_items
(id, order_id, product_id, quantity, unit_price)
VALUES (?, ?, ?, ?, ?)
""", order_items)


# ============================================================
# 10. REVIEWS
# ============================================================

review_texts = {

    5: [
        "Excellent product. Highly recommended.",
        "Amazing quality and great value.",
        "Very happy with the purchase.",
        "Works perfectly."
    ],

    4: [
        "Good product and worth the money.",
        "Very good overall.",
        "Quality is good.",
        "Happy with the purchase."
    ],

    3: [
        "Average product.",
        "It is okay for the price.",
        "Could be improved.",
        "Decent but not exceptional."
    ],

    2: [
        "Not very satisfied.",
        "Quality could be better.",
        "Expected more from this product."
    ],

    1: [
        "Very disappointing.",
        "Product stopped working quickly.",
        "Not worth the price."
    ]
}

reviews = []

for review_id in range(1, 2501):

    product_id = random.randint(1, len(products))
    customer_id = random.randint(1, 500)

    rating = random.choices(
        [1, 2, 3, 4, 5],
        weights=[5, 5, 15, 30, 45],
        k=1
    )[0]

    reviews.append((
        review_id,
        product_id,
        customer_id,
        rating,
        random.choice(review_texts[rating]),
        (
            datetime.now()
            - timedelta(days=random.randint(0, 365))
        ).strftime("%Y-%m-%d")
    ))

cursor.executemany("""
INSERT INTO reviews
(id, product_id, customer_id, rating, review_text, review_date)
VALUES (?, ?, ?, ?, ?, ?)
""", reviews)

conn.commit()


# ============================================================
# 11. DATABASE SUMMARY
# ============================================================

print("\n" + "=" * 70)
print("📊 DATABASE SUMMARY")
print("=" * 70)

for table in [
    "customers",
    "products",
    "orders",
    "order_items",
    "reviews"
]:

    count = pd.read_sql_query(
        f"SELECT COUNT(*) AS count FROM {table}",
        conn
    ).iloc[0]["count"]

    print(f"{table:<15}: {count:,}")


# ============================================================
# 12. GEMINI MODEL
#
# NOTE:
# Do NOT pass temperature.
# Gemini 3.6 Flash uses fixed sampling defaults.
# ============================================================

llm = ChatGoogleGenerativeAI(
    model="gemini-3.6-flash",
    google_api_key=os.environ["GEMINI_API_KEY"]
)

print("\n✅ Gemini 3.6 Flash connected")


# ============================================================
# 13. IMPORTANT GEMINI RESPONSE HANDLER
# ============================================================

def extract_text(response):

    content = response.content

    # Case 1:
    # Normal string response
    if isinstance(content, str):
        return content.strip()

    # Case 2:
    # Gemini returns a list of content blocks
    if isinstance(content, list):

        text_parts = []

        for item in content:

            # Dictionary content block
            if isinstance(item, dict):

                text = item.get("text")

                if text:
                    text_parts.append(str(text))

            # Plain string block
            elif isinstance(item, str):

                text_parts.append(item)

        return "\n".join(
            text_parts
        ).strip()

    # Fallback
    return str(content).strip()


# ============================================================
# 14. DATABASE SCHEMA
# ============================================================

def get_schema():

    schema = ""

    tables = pd.read_sql_query("""
        SELECT name
        FROM sqlite_master
        WHERE type='table'
        AND name NOT LIKE 'sqlite_%'
    """, conn)

    for table in tables["name"]:

        columns = pd.read_sql_query(
            f"PRAGMA table_info({table})",
            conn
        )

        schema += f"\nTABLE: {table}\n"

        schema += ", ".join(
            columns["name"].tolist()
        )

        schema += "\n"

    return schema


SCHEMA = get_schema()


# ============================================================
# 15. GENERATE SQL
# ============================================================

def generate_sql(question):

    prompt = f"""
You are an expert SQLite e-commerce data analyst.

DATABASE SCHEMA:

{SCHEMA}


RELATIONSHIPS:

orders.customer_id = customers.id

order_items.order_id = orders.id

order_items.product_id = products.id

reviews.product_id = products.id

reviews.customer_id = customers.id


BUSINESS DEFINITIONS:

Revenue:

SUM(
    order_items.quantity *
    order_items.unit_price
)


Profit:

SUM(
    order_items.quantity *
    (
        order_items.unit_price -
        products.cost
    )
)


Completed sales:

orders.status = 'completed'


Cancelled:

orders.status = 'cancelled'


Returned:

orders.status = 'returned'


Pending:

orders.status = 'pending'


USER QUESTION:

{question}


TASK:

Generate ONE SQLite SELECT query that answers
the user's question.


RULES:

1. Return ONLY SQL.
2. Do not use markdown.
3. Only SELECT or WITH queries.
4. Never INSERT.
5. Never UPDATE.
6. Never DELETE.
7. Never DROP.
8. Never ALTER.
9. Never CREATE.
10. Never TRUNCATE.
11. Never modify the database.
12. Use only actual columns.
13. Do not invent data.
14. Revenue = quantity * unit_price.
15. Profit = quantity * (unit_price - cost).
16. Use completed orders for revenue/profit unless
    the user explicitly asks otherwise.

SQL:
"""

    response = llm.invoke(prompt)

    # FIX:
    # Gemini content may be a list
    sql = extract_text(response)

    # Remove markdown fences
    sql = re.sub(
        r"```sql",
        "",
        sql,
        flags=re.IGNORECASE
    )

    sql = re.sub(
        r"```",
        "",
        sql
    )

    # Remove possible "SQL:" prefix
    sql = re.sub(
        r"^\s*SQL\s*:\s*",
        "",
        sql,
        flags=re.IGNORECASE
    )

    return sql.strip()


# ============================================================
# 16. SQL SAFETY
# ============================================================

def safe_sql(sql):

    sql_upper = sql.strip().upper()

    if not (
        sql_upper.startswith("SELECT")
        or sql_upper.startswith("WITH")
    ):
        return False

    blocked = [
        "INSERT",
        "UPDATE",
        "DELETE",
        "DROP",
        "ALTER",
        "TRUNCATE",
        "CREATE",
        "REPLACE",
        "ATTACH",
        "DETACH"
    ]

    for word in blocked:

        if re.search(
            rf"\b{word}\b",
            sql_upper
        ):
            return False

    return True


# ============================================================
# 17. EXECUTE SQL
# ============================================================

def run_sql(sql):

    if not safe_sql(sql):

        raise ValueError(
            "Unsafe SQL detected. "
            "Only SELECT queries are allowed."
        )

    return pd.read_sql_query(
        sql,
        conn
    )


# ============================================================
# 18. ANALYZE RESULT
# ============================================================

def analyze_result(
    question,
    sql,
    result
):

    if result.empty:

        result_text = "No rows returned."

    else:

        result_text = result.to_string(
            index=False
        )

    prompt = f"""
You are an expert e-commerce business analyst.

USER QUESTION:

{question}


SQL QUERY:

{sql}


DATABASE RESULT:

{result_text}


TASK:

Answer the user's question using ONLY
the database result.

RULES:

1. Do not invent numbers.
2. Do not invent products.
3. Do not invent customers.
4. Use ₹ for monetary values.
5. Clearly answer the question.
6. Explain important comparisons.
7. Keep the response concise.
8. Do not generate SQL.
9. Do not discuss internal implementation.


ANSWER:
"""

    response = llm.invoke(prompt)

    # FIX:
    # Gemini content may be a list
    return extract_text(response)


# ============================================================
# 19. MAIN ANALYST
# ============================================================

def ask_analyst(question):

    try:
        print("\n🔄 Processing your question...")

        # ====================================================
        # SPECIAL CASE: CITY WITH MOST CUSTOMERS
        # ====================================================

        q = question.lower().strip()

        if (
            "which city" in q
            and "most customers" in q
        ):

            sql = """
            SELECT
                city,
                COUNT(*) AS customer_count
            FROM customers
            GROUP BY city
            ORDER BY customer_count DESC
            LIMIT 1
            """

            print("\n🔎 SQL:")
            print(sql)

            result = pd.read_sql_query(
                sql,
                conn
            )

            print("\n📊 RESULT:")
            display(result)

            if not result.empty:

                city = result.iloc[0]["city"]
                count = result.iloc[0]["customer_count"]

                return (
                    f"🏙️ {city} has the most customers "
                    f"with {count:,} customers."
                )

            return "No customer data found."


        # ====================================================
        # NORMAL GEMINI → SQL → SQLITE → GEMINI FLOW
        # ====================================================

        print("\n🤖 Asking Gemini to generate SQL...")

        sql = generate_sql(question)

        print("\n🔎 GENERATED SQL")
        print("-" * 70)
        print(sql)

        if not safe_sql(sql):

            return (
                "❌ Unsafe SQL generated. "
                "Only read-only queries are allowed."
            )

        print("\n⚙️ Executing SQL...")

        result = run_sql(sql)

        print("\n📊 DATABASE RESULT")
        print("-" * 70)

        display(result)

        print("\n🧠 Generating analysis...")

        answer = analyze_result(
            question,
            sql,
            result
        )

        return answer

    except Exception as e:

        print("\n❌ ERROR")
        print("-" * 70)
        print(
            type(e).__name__,
            ":",
            str(e)
        )

        return f"❌ Error: {str(e)}"
# ============================================================
# 20. TEST
# ============================================================

print("\n" + "=" * 70)
print("🧪 TESTING AI ANALYST")
print("=" * 70)

test_question = (
    "Which products generated the most revenue?"
)

print(
    "\n🧑 Question:",
    test_question
)

test_answer = ask_analyst(
    test_question
)

print("\n💡 AI ANSWER")
print("-" * 70)
print(test_answer)


# ============================================================
# 21. INTERACTIVE CHAT
# ============================================================

print("\n" + "=" * 70)
print("🤖 E-COMMERCE AI DATA ANALYST READY")
print("=" * 70)

print("""
Ask questions such as:

• Which products generated the most revenue?

• What is our total revenue?

• What is our total profit?

• Which category generates the most revenue?

• Which products are the most profitable?

• What is our average order value?

• How many orders were cancelled?

• What is our cancellation rate?

• Which city has the most customers?

• Which products have low inventory?

• Compare Electronics and Fashion revenue.

• Which products have the highest profit margin?

• What are the worst-rated products?

Type 'exit' to quit.
""")


while True:

    try:

        question = input(
            "\n🧑 You: "
        )

    except EOFError:

        break

    if question.strip().lower() in [
        "exit",
        "quit",
        "q"
    ]:

        print(
            "\n👋 Goodbye!"
        )

        break

    if not question.strip():

        continue

    print(
        "\n🤖 AI Analyst"
    )

    print(
        "=" * 70
    )

    answer = ask_analyst(
        question
    )

    print(
        "\n💡 ANALYSIS"
    )

    print(
        "-" * 70
    )

    print(
        answer
    )

    print(
        "=" * 70
    )
````
